```python
# Install required packages
!pip install pillow

# Import necessary modules
from IPython.display import display
from PIL import Image, ImageOps, UnidentifiedImageError # Import UnidentifiedImageError
import io
import IPython
import ipywidgets as widgets

# Upload widget
upload_btn = widgets.FileUpload(
    accept='image/*',
    multiple=False
)

# Display upload button
display(upload_btn)

def handle_upload(change):
    if upload_btn.value:
        # Get uploaded image content
        for filename, file_info in upload_btn.value.items():
            image_data = file_info['content']
            try:
                image = Image.open(io.BytesIO(image_data))
            except UnidentifiedImageError:
                print(f"Error: Could not identify image format for {filename}. Please upload a valid image file (e.g., .png, .jpg, .jpeg).")
                return # Stop processing if image format is not recognized

            print(f"Original image: {filename}")
            display(image)

            # Example processing: convert to grayscale
            gray_image = ImageOps.grayscale(image)

            print("Grayscale image:")
            display(gray_image)

# Trigger image handling on upload
upload_btn.observe(handle_upload, names='value')
```

OpenCV (cv2)
Purpose: Core library for image and video processing.

Functions used:

cv2.imread() – Load an image.

cv2.cvtColor() – Convert between color spaces (e.g., BGR to Grayscale).

cv2.GaussianBlur() – Apply Gaussian smoothing.

cv2.Canny() – Perform edge detection.

cv2.threshold() – Apply binary thresholding.

(Optional: cv2.imwrite() – Save image to file.)

NumPy (numpy or np)
Purpose: Used for numerical operations and array handling.

Role in this code: While it's imported, it's not directly used in the current snippet, but it's often used to manipulate image arrays or generate filters/masks.

Matplotlib (matplotlib.pyplot or plt)
Purpose: Visualization and plotting of images.

Functions used:

plt.figure() – Create a new figure for plotting.

plt.subplot() – Organize plots in a grid layout.

plt.imshow() – Display an image.

plt.title() – Add a title to each subplot.

plt.axis('off') – Hide axis lines and labels.

plt.tight_layout() – Optimize spacing between subplots.

plt.show() – Display the figure.
